```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_moons
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
                     

X, y = make_moons(n_samples=500, noise=0.30, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
                                        

rnd_clf = RandomForestClassifier(n_estimators=5, max_leaf_nodes=5, random_state=42)
rnd_clf.fit(X_train, y_train)

y_pred_rf = rnd_clf.predict(X_test)
print("La exactitud de la prediccion es:")
accuracy_score(y_test, y_pred_rf)
```

```python
print(classification_report(y_test, y_pred_rf))
```

## Solución Parte III — Resultados y tabla de exactitud
Ejecuta esta celda para completar los 15 casos con el mismo conjunto de prueba.

```python
import pandas as pd
resultados=[]
for n in [5,50,100,500,1000]:
    fila={'n_estimators':n}
    for hojas in [5,8,16]:
        rf=RandomForestClassifier(n_estimators=n,max_leaf_nodes=hojas,random_state=42,n_jobs=-1)
        rf.fit(X_train,y_train)
        fila[f'max_leaf_nodes={hojas}']=round(100*accuracy_score(y_test,rf.predict(X_test)),2)
    resultados.append(fila)
tabla=pd.DataFrame(resultados).set_index('n_estimators')
print('Exactitud inicial (5 árboles y 5 hojas):',f'{100*accuracy_score(y_test,y_pred_rf):.2f}%')
print('Tabla de exactitudes (%)')
display(tabla)
```

**Conclusión:** Más árboles generalmente estabilizan la predicción, aunque la exactitud no siempre aumenta. Más hojas permiten mayor complejidad, pero también pueden provocar sobreajuste. Los porcentajes dependen de la partición de datos y se calculan al ejecutar el notebook.
